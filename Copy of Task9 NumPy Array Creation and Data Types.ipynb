```python
import numpy as np

arr = np.array([[10, 20, 30],
                [40, 50, 60]])

print("Shape:", arr.shape)
print("Dimensions:", arr.ndim)
print("Size:", arr.size)
print("Data type:", arr.dtype)
```

```python
import numpy as np

a = np.array([10, 20, 30])
b = np.array([1.5, 2.5, 3.5])
c = np.array([True, False, True])

print(a.dtype)
print(b.dtype)
print(c.dtype)
```

```python
import numpy as np

a = np.array([10, 20, 30])
b = np.array([1, 2, 3])

print("Addition:", a + b)
print("Subtraction:", a - b)
print("Multiplication:", a * b)
print("Division:", a / b)
```

```python
import numpy as np

my_list = [10, 20, 30]
my_array = np.array([10, 20, 30])

print("Python list:", my_list)
print("NumPy array:", my_array)
print("Array multiplied by 2:", my_array * 2)
```
